For this homework, I chose a book "Le avventure di Pinocchio" by Carlo Collodi. Since I study Italian language and culture, I wanted to work with an Italian literary text.

My goal is to analyse the text and create a word cloud to see which words appear most frequently.

```python
import urllib.request
text = urllib.request.urlopen("https://raw.githubusercontent.com/GITenberg/Le-avventure-di-Pinocchio-Storia-di-un-burattino_52484/master/52484-0.txt").read().decode("utf-8")

text[:90]


```

```python
words = text.split( )
print(words[:10])
print(words[-10:])
print(len(words))
```

```python
long_words = []
for word in words:
    if len(word) > 10:
        long_words.append(word)
print(long_words[:15])
```

I used "urllib.request" to read the text of "Pinocchio" from an online file. Then I used "text[:90]" to display the first 90 characters.

Next, I used "split()" to divide the text into words. I displayed the first and last 10 words and used "len()" to count them. I also used a for loop to find words that have more than 10 characters. I saved these words in a new list and displayed the first 15 results.

```python
from wordcloud import WordCloud
WordCloud().generate_from_text(text).to_image()
```

I used the WordCloud library to create a word cloud from the text of "Pinocchio". The most frequent words appear larger, while less frequent words appear smaller.

```python

stopwords_list = ["e", "ed", "il", "la", "lo", "i", "gli", "le", "un", "una", "uno", "di", "da", "a", "in", "con", "per", "che", "non", "si", "ma", "era", "come", "del", "della", "al", "alla", "è", "sono", "ha", "ho", "mi", "ti", "ci"]

WordCloud(stopwords=stopwords_list,background_color="white").generate_from_text(text).to_image()
```

```python

stopwords_list += ["perché","perchè" "quando", "più", "allora",
                   "disse", "quel", "così", "poi",
                   "aveva", "tutti", "tutto", "me",
                   "se", "chi", "due", "subito",
                   "Project", "Gutenberg", "and", "mai", "dalla", "cosa", "dopo"]

WordCloud(stopwords=stopwords_list, background_color="white").generate_from_text(text).to_image()
```

In the first word cloud, the most common words were mainly Italian articles, prepositions and conjunctions, such as "che", "il" and "una". I added a list of Italian stopwords to remove some of these words and make the word cloud more meaningful.

In the final word cloud, words such as "Pinocchio", "burattino", "Geppetto" and "Fata" are clearly visible. These words represent important characters in the story.

I noticed that removing stopwords made it easier to identify the main characters and themes of the text. However, some common words still appear, which shows that the selection of stopwords affects the result.

```python

def color_function(word, font_size, position, orientation, random_state=None, **additional_args):
    if len(word) < 5:
        return "gold"
    elif len(word) < 9:
        return "royalblue"
    else:
        return "firebrick"

WordCloud(stopwords=stopwords_list,background_color="white", width=800,height=400).generate_from_text(text).recolor(color_func=color_function).to_image()
```

I  experimented with the colours of the word cloud. I used a custom colour function to give different colours to words based on their length. Short words are yellow, medium-length words are blue and longer words are red.

This makes the word cloud more colourful and allows me to compare word lengths visually.
