```python
import pandas as pd
df = pd.read_csv("Italian authors.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Author", y="Lifespan", legend=False)
plt.title("Lifespan of Italian Authors")
plt.ylabel("Lifespan (years)")
plt.xlabel("Author")
plt.show()
```

The chart shows that the lifespans of the Italian authors vary considerably. Alessandro Manzoni had the longest lifespan at 88 years, while Giacomo Leopardi had the shortest lifespan at 39 years.

```python
avg_lifespan = df.groupby("Literary period")["Lifespan"].mean()
avg_lifespan.plot(kind="bar", legend=False)
plt.title("Average Lifespan by Literary Period")
plt.ylabel("Average lifespan (years)")
plt.xlabel("Literary period")
plt.show()
```

The chart shows that authors from the Realism period had the highest average lifespan. Authors from the Renaissance period had the lowest average lifespan.

```python
plt.scatter(df["Birth year"], df["Lifespan"])
plt.xlabel("Birth year")
plt.ylabel("Lifespan")
plt.title("Birth Year vs Lifespan")
plt.show()
```

The scatter plot does not show a very strong relationship between birth year and lifespan. Some authors born in later years lived longer, but there are also several exceptions.

```python
df[["Lifespan"]].boxplot()
plt.title("Distribution of Authors' Lifespans")
plt.ylabel("Lifespan (years)")
plt.show()
```

The boxplot shows that the median lifespan of the authors is about 67 years. Most authors lived between approximately 56 and 82 years.
