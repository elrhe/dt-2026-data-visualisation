Create a chart showing who puts in the most effort. Effort can be find from number of hours studied. (For example Student A has the highest study hours meaning they put in the most effort.)

```python
import pandas as pd
df= pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt 
avg_hours =df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Students", y="Hours_Studied", color= "#301934")
#colors=plt.cm.plasma(range(len(avg hours)))
plt.title("Average study hours per students")
plt.ylabel("Hour studied")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
#s=df["Hours_Studied"]*30 
#alpha=0.6 makes the bubbles bit transparent
plt.xlabel("Hours studied")
plt.ylabel("Grade")
plt.title("Grade vs hours studied")
```

```python
plt.Figure(figsize=(8,5))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot for hours studied and grades")
plt.show
```
